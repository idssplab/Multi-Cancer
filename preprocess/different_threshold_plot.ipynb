```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set the style
sns.set_style('darkgrid')
# Set the color palette
sns.set_palette('crest')

# Data from the provided tables
tests = ['BRCA', 'LUAD', 'COAD']

# Results at 1 year cutoff
prc_1yr = [0.25567, 0.55152, 0.47506]
auc_1yr = [0.47034, 0.84834, 0.88535]
c_index_1yr = [0.69123, 0.66439, 0.75839]

# Results at 3 years cutoff
prc_3yr = [0.51116, 0.73728, 0.52850]
auc_3yr = [0.74806, 0.83310, 0.72333]
c_index_3yr = [0.74727, 0.72667, 0.71546]


# Results at 5 years cutoff
prc_5yr = [0.348, 0.509, 0.712]
auc_5yr = [0.839, 0.645, 0.712]
c_index_5yr = [0.823, 0.587, 0.696]

# Results at 7 years cutoff
prc_7yr = [0.99626, 0.99979, 1.00000]
auc_7yr = [0.90145, 0.64387, 0.61800]
c_index_7yr = [0.35370, 0.50053, 0.49487]

# Data for plotting
prc_data = [prc_1yr, prc_3yr, prc_5yr, prc_7yr]
auc_data = [auc_1yr, auc_3yr, auc_5yr, auc_7yr]
c_index_data = [c_index_1yr, c_index_3yr, c_index_5yr, c_index_7yr]
x = np.arange(len(tests))  # the label locations
width = 0.2  # the width of the bars

fig, ax = plt.subplots(3, 1, figsize=(10, 15))

# PRC plot
for i, (prc, label) in enumerate(zip(prc_data, ['1 year', '3 years', '5 years', '7 years'])):
    ax[0].bar(x + i * width - width, prc, width, label=label)
ax[0].set_title('PRC Comparison at Different Time Thresholds')
ax[0].set_xticks(x)
ax[0].set_xticklabels(tests)
ax[0].set_yticks(np.arange(0, 1.1, 0.1)) 
ax[0].set_ylabel('PRC')
ax[0].legend()

# AUC plot
for i, (auc, label) in enumerate(zip(auc_data, ['1 year', '3 years','5 years',  '7 years'])):
    ax[1].bar(x + i * width - width, auc, width, label=label)
ax[1].set_title('AUC Comparison at Different Time Thresholds')
ax[1].set_xticks(x)
ax[1].set_xticklabels(tests)
ax[1].set_yticks(np.arange(0, 1.1, 0.1)) 
ax[1].set_ylabel('AUC')
ax[1].legend()

# C-Index plot
for i, (c_index, label) in enumerate(zip(c_index_data, ['1 year', '3 years', '5 years', '7 years'])):
    ax[2].bar(x + i * width - width, c_index, width, label=label)
ax[2].set_title('C-Index Comparison at Different Time Thresholds')
ax[2].set_xticks(x)
ax[2].set_xticklabels(tests)
ax[2].set_yticks(np.arange(0, 1.1, 0.1)) 
ax[2].set_ylabel('C-Index')
ax[2].legend()

plt.tight_layout()
#save as pdf 
plt.savefig('metrics_comparison_different_thresholds.pdf')
plt.show()
```
