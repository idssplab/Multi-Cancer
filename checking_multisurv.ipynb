```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
multisurv_labels = pd.read_csv('multisurv_labels.tsv', sep='\t', index_col=0)
multisurv_labels.head()
```

```python

# select only the cancer types BRCA, LUAD, COAD
cancers = ['BRCA', 'LUAD', 'COAD']
selected_sets = multisurv_labels[multisurv_labels['project_id'].isin(cancers)]
selected_sets.head()
```

```python
# Calculate the number of positive samples in event for each cancer type
positive_samples = selected_sets.groupby(['project_id', 'event', 'group']).size().unstack()
positive_samples

```

```python
# I want all the submitter_id from the test set
selected_sets_test = selected_sets[selected_sets['group'] == 'test']
selected_sets_test.head()
```

```python
# I want to select all the uuuids from Data/TCGA/BRCA, LUAD, COAD that contain these submitter_ids
import os
import json

# Get all the uuids from the test set
uuids_brca = selected_sets_test[selected_sets_test['project_id'] == 'BRCA'].index
uuids_luad = selected_sets_test[selected_sets_test['project_id'] == 'LUAD'].index
uuids_coad = selected_sets_test[selected_sets_test['project_id'] == 'COAD'].index
```

```python
path = 'Data/TCGA/'
# create table with all the uuids and sumitter_ids
uuids = []
submitter_ids = []
for cancer in cancers:
    with open(path + cancer + '/meta/clinical.json') as f:
        data = json.load(f)
        for entry in data:
            uuids.append(entry['uuid'])
            submitter_ids.append(entry['submitter_id'])
```
